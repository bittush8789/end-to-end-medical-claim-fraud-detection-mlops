# Exploratory Data Analysis (EDA) - Medical Claim Fraud Detection
This notebook explores the synthetic raw insurance claims dataset to understand features, distribution, missing values, correlations, and fraud patterns.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load data
df = pd.read_csv('../data/raw/claims_data.csv')
print(f'Dataset Shape: {df.shape}')
```

## Basic Statistics & Structure


```python
df.info()
df.describe()
```

## Missing Value Analysis


```python
missing = df.isnull().sum()
print('Missing values per column:')
print(missing[missing > 0])
```

## Target Variable (is_fraud) Distribution


```python
plt.figure(figsize=(6, 4))
sns.countplot(x='is_fraud', data=df, palette='Set2')
plt.title('Distribution of Fraudulent vs Legitimate Claims')
plt.xlabel('Is Fraud')
plt.ylabel('Count')
plt.show()
print(df['is_fraud'].value_counts(normalize=True))
```

## Correlation Analysis of Numerical Features


```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
plt.figure(figsize=(12, 10))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

## Key Fraud Patterns
Let's look at claim amounts vs fraud status, and provider fraud histories.


```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='is_fraud', y='claim_amount', data=df, palette='Set2')
plt.title('Claim Amount Distribution by Fraud Status')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='is_fraud', y='previous_fraud_cases', data=df, palette='Set2')
plt.title('Provider Previous Fraud Cases by Fraud Status')
plt.show()
```
