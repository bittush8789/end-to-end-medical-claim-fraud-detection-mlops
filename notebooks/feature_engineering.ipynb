# Feature Engineering - Medical Claim Fraud Detection
This notebook covers the creation of derived interactive features that capture indicators of fraudulent claims.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('../src')
from feature_engineering import FeatureEngineer

df = pd.read_csv('../data/raw/claims_data.csv')
fe = FeatureEngineer()
df_engineered = fe.add_features(df)
print(f'Columns added: {[c for c in df_engineered.columns if c not in df.columns]}')
```

## Analyzing Engineered Features


```python
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df_engineered, x='claim_to_coverage_ratio', hue='is_fraud', fill=True, common_norm=False)
plt.title('Distribution of Claim-to-Coverage Ratio by Fraud Status')
plt.xlim(0, 2)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_engineered, x='is_fraud', y='risk_score', palette='Set1')
plt.title('Risk Score by Fraud Status')
plt.show()
```
